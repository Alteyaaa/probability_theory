* Рассмотрим событие $A$ - все 5 книг на полке стоят по порядку.

Все возможные перестановки 5 книг на полке - $P_5 = 5!$. Благоприятный исход 1.

$P(A) = \frac{1}{5!}$

```python
round(1/sc.factorial(5), 4)
```

* Рассмотрим событие $A$ - все цифры в номере различны.

Общее число исходов - $9 \cdot \bar A_{10}^4 = 9 \cdot 10^4$. Благоприятные исходы - размещения без повторения $9 \cdot A_{9}^4 = 9 \cdot \frac{9!}{6!}$.

$P(A) = \frac{9 \cdot  A_{9}^4}{9 \cdot \bar A_{10}^4} = \frac{A_{9}^4}{\bar A_{10}^4}$

```python
round(sc.perm(9, 4)/10**4, 4)
```

* Рассмотрим событие $B$ - все цифры номера одинаковые.

Общее число исходов - $9 \cdot \bar A_{10}^4 = 9 \cdot 10^4$. Всего номеров с одинаковыми цифрами может быть 9, учитывая то, что первый не ноль.

$P(B) = \frac{9}{9 \cdot \bar A_{10}^4} = \frac{1}{\bar A_{10}^4}$

```python
round(1/10**4, 4)
```

* Рассмотрим событие $C$ - все цифры номера нечетные.

Общее число исходов - $9 \cdot \bar A_{10}^4 = 9 \cdot 10^5$. Из 10 цифр всего 5 нечетных. Они могут повторяться, первое число априори не 0, потому что это не нечетная цифра. Следовательно, благоприятные исходы - $\bar A_5^5 = 5^5$.

$P(C) = \frac{\bar A_5^5}{9 \cdot \bar A_{10}^4}$

```python
round(5**5/ (9 * 10**4), 4)
```

* Рассмотрим событие $A$- выпало 11 очков.

Общее число возможных исходов определяется как размещения с повторениями - $\bar A_{6}^3$. Благоприятные исходы складываются из нескольких случаев:



1.   {6, 4, 1}, {5, 4, 2}, {2, 3, 6} - 3 варианта с 3! перестановками = 18
2.   {5, 5, 1}, {3, 4, 4}, {5, 3, 3} - 3 варианта с 3 перестановками = 9

Всего 27 благоприятных исходов. Тогда

$P(A) = \frac{27}{\bar A_6^3}$



```python
round(27/6**3, 4)
```
