* Рассмотрим событие $A$ - по каждой номинации установлены различные призы.

В таком случае каждый из 10 кинофильмов может выиграть от 1 до 5 номинаций (имеются повторения).

$\bar{A}_{10}^{5} = 10^5$

```python
n = 10
k = 5
round(n**k, 4)
```

В этом случае порядок получения наград не имеет значения, т.к. они все одинаковые.

$\bar{C}_{10}^{5} = \frac{14!}{5! \cdot 9!}$

```python
n = 10
k = 5
round(sc.comb(n+k-1, k), 4)
```

* Рассмотрим событие $A$ - ребенок взял три карточки и сложил слово "ТОР".

Всего имеется 6 карточек с буквами, были использованы 3. Порядок важен.

$A_{6}^{3} = \frac{6!}{3!}$

Слово "ТОР" является одной из $A_{6}^{3}$ возможных комбинаций.

$P(A) = \frac{1}{A_{6}^3}$


```python
n = 6
k = 3
round(1/sc.perm(n, k), 4)
```

* Рассмотрим событие $B$ - ребенок взял все 6 карточек и сложил слово "ТЕОРИЯ".

Всего имеется 6 карточек с буквами, были использованы 6

$P_{6} = 6!$

Слово "ТЕОРИЯ" является одной из $P_{6}$ возможных комбинаций.

$P(B) = \frac{1}{P_{6}}$

```python
n = 6
round(1/sc.factorial(n), 4)
```

* Рассмотрим событие $A$ - из 6-ти карточек с буквами "А"(3), "Н"(2) и "С"(1) было составлено слово "АНАНАС".

Было использовано 6 карточек, в которых имелись повторяющиеся буквы.

$\bar{P}_{6} = \frac{6!}{3! \cdot 2!}$

Слово "АНАНАС" является одной из $\bar{P}_{6}$ возможных комбинаций.

$P(A) = \frac{1}{\bar{P}_{6}}$

```python
n = 6
n1 = 3
n2 = 2
round((sc.factorial(n1)*sc.factorial(n2))/sc.factorial(n), 4)
```

* Рассмотрим событие $A$ - никакие 2 лица одного пола не сели рядом.

Общее число перестановок - $10!$. Благоприятных исходов - $2 \cdot 5! \cdot 5!$, поскольку мужчины и женщины должны сесть чередуясь и могут меняться местами между своим полом. Следовательно, они могут сесть на четные и нечетные места по очереди и осуществять свои перестановки среди людей одного пола.

$P(A) = \frac{2 \cdot 5! \cdot 5!}{n!}$

```python
n1 = 5
n2 = 5
n = 10
round((2*sc.factorial(5)*sc.factorial(5))/sc.factorial(10), 4)
```

* Рассмотрим событие $B$ - супруги сели рядом, при условии, что мужчины и женщины образуют 5 супружеских пар.

Общее число перестановок - $10!$. Благоприятных исходов - $5!*2^5$. Число благоприятных исходов объясняется тем, что пары могут осуществлять перестановки между собой ($5!$), и в каждой паре муж и жена так же могут меняться местами ($2!=2$).

$P(B) = \frac{5! \cdot 2^5}{10!}$

```python
n1 = 5
n = 10
round((sc.factorial(n1)*(2**n1))/sc.factorial(n), 4)
```

* Рассмотрим событие $A$ - пассажиры попали в 2 купе.

Всего в вагоне 36 мест, людей 7. Таким образом, всего они могут разместиться $C_{36}^7$ вариантами. Благоприятные исходы составляют все варианты выбора 2 купе из 9 - $C_9^2$ и все варианты рассадок 7 людей по 8 местам из двух купе - $C_8^7$.

$P(A) = \frac{C_9^2 \cdot C_8^7}{C_{36}^7}$

```python
round((sc.comb(9, 2)*sc.comb(8, 7))/sc.comb(36, 7), 5)
```

> **Замечание.** Ответ был приведён с округлением до 5 знаков после запятой, поскольку при округлении до 4 получается ноль.

* Рассмотрим событие $B$ - пассажиры попали в 7 купе.

Общее число размещений 7 людей среди 36 мест - $C_{36}^7$. Благоприятные исходы составляют все варианты выбора 7 купе из 9 - $C_{9}^7$ и все варианты размещения 7 пассажиров среди 28 мест с учетом того, что в купе должен быть хотя бы 1 человек - ${С_4^1}^7 = 4^7$.

$P(B) = \frac{C_{9}^7 \cdot 4^7}{C_{36}^7}$

```python
round((sc.comb(9, 7)*4**7)/sc.comb(36, 7), 4)
```

* Рассмотрим событие $C$ - пассажиры попали в 3 купе.

Общее число размещений 7 людей среди 36 мест — $C_{36}^{7}$. Благоприятные исходы составляют все варианты выбора 3 купе из 9 — $C_{9}^{3}$ и все варианты размещения 7 пассажиров среди 12 мест (в одном купе хотя бы 1 пассажир). Последнее достигается при следующем размещении пассажиров: "4+2+1", "2+2+3" и "3+3+1". Схема "4+2+1" даёт $3!$ перестановок ролей купе и $C_{4}^{4} \cdot C_{4}^{2} \cdot C_{4}^{1}$ способов на каждую; схема "2+2+3" даёт $\frac{3!}{2!}$ перестановки и $C_{4}^{2} \cdot C_{4}^{2} \cdot C_{4}^{3}$ способов; схема "3+3+1" даёт $\frac{3!}{2!}$ перестановки и $C_{4}^{3} \cdot C_{4}^{3} \cdot C_{4}^{1}$ способов.

$$P(C) = \frac{C_{9}^{3}\left(6 \cdot C_{4}^{4} C_{4}^{2} C_{4}^{1} + 3 \cdot C_{4}^{2} C_{4}^{2} C_{4}^{3} + 3 \cdot C_{4}^{3} C_{4}^{3} C_{4}^{1}\right)}{C_{36}^{7}}$$

```python
pos1 = sc.comb(4, 1)* sc.comb(4, 2)
pos2 = sc.comb(4, 2)* sc.comb(4, 2) * sc.comb(4, 3)
pos3 = sc.comb(4, 1)* sc.comb(4, 3) * sc.comb(4, 3)
pos4 = sc.comb(9, 3)
round((6*pos1+3*pos2+3*pos3)*pos4/sc.comb(36, 7), 4)
```

* Рассмотрим событие $A$ - все цифры в номере различны.

Общее число исходов - $9 \cdot \bar A_{10}^4 = 9 \cdot 10^4$. Благоприятные исходы - размещения без повторения $9 \cdot A_{9}^4 = 9 \cdot \frac{9!}{6!}$.

$P(A) = \frac{9 \cdot  A_{9}^4}{9 \cdot \bar A_{10}^4} = \frac{A_{9}^4}{\bar A_{10}^4}$

```python
round(sc.perm(9, 4)/10**4, 4)
```

* Рассмотрим событие $B$ - все цифры номера одинаковые.

Общее число исходов - $9 \cdot \bar A_{10}^4 = 9 \cdot 10^4$. Всего номеров с одинаковыми цифрами может быть 9, учитывая то, что первый не ноль.

$P(B) = \frac{9}{9 \cdot \bar A_{10}^4} = \frac{1}{\bar A_{10}^4}$

```python
round(1/10**4, 4)
```

* Рассмотрим событие $C$ - все цифры номера нечетные.

Общее число исходов - $9 \cdot \bar A_{10}^4 = 9 \cdot 10^5$. Из 10 цифр всего 5 нечетных. Они могут повторяться, первое число априори не 0, потому что это не нечетная цифра. Следовательно, благоприятные исходы - $\bar A_5^5 = 5^5$.

$P(C) = \frac{\bar A_5^5}{9 \cdot \bar A_{10}^4}$

```python
round(5**5/ (9 * 10**4), 4)
```

# Тема 4. Полная вероятность и формула Байеса

```python
import scipy.special as sc
```

В одной урне 15 белых и 5 черных шаров, а в другой - 6 белых и 1 черный шар. Из первой урны случайным образом вынимают 3 шара и опускают во вторую урну. После этого из второй урны также случайно вынимают 1 шар. Найти вероятность того, что шар, вынутый из второй урны - белый.

* Рассмотрим событие $A$ - вынутый из второй урны шар - белый.

Изначально в первой урне 20 шаров, а во второй 7. После перекладывания из первой урны во вторую 3 шаров в первой урне остается 17 шаров, а во второй становится 10. Затем из второй урны вынимают один шар и там остается 9 шаров.
Белый шар мог быть вынут из второй корзины, в которой от 6 до 9 белых шаров, в зависимости от того, сколько каких шаров было переложено в первом действии. Таким образом, может получиться 3 ситуации - из первой корзины было переложено:

3 белых шара - $\dfrac{15}{20}\cdot\dfrac{14}{19}\cdot\dfrac{13}{18}$

2 белых шара и 1 черный - $3\cdot\dfrac{15}{20}\cdot\dfrac{14}{19}\cdot\dfrac{5}{18}$

1 белый шар и 2 черных - $3\cdot\dfrac{15}{20}\cdot\dfrac{5}{19}\cdot\dfrac{4}{18}$

3 черных шара - $\dfrac{5}{20}\cdot\dfrac{4}{19}\cdot\dfrac{3}{18}$

Затем, взять 1 белый шар из второй корзины можно как $\dfrac{9}{10}$, $\dfrac{8}{10}$, $\dfrac{7}{10}$, $\dfrac{6}{10}$ соответственно. Тогда общий ответ будет образовываться как

$P(A) = \dfrac{15}{20}\cdot\dfrac{14}{19}\cdot\dfrac{13}{18}\cdot\dfrac{9}{10} + 3\cdot\dfrac{15}{20}\cdot\dfrac{14}{19}\cdot\dfrac{5}{18}\cdot\dfrac{8}{10} + 3\cdot\dfrac{15}{20}\cdot\dfrac{5}{19}\cdot\dfrac{4}{18}\cdot\dfrac{7}{10} + \dfrac{5}{20}\cdot\dfrac{4}{19}\cdot\dfrac{3}{18}\cdot\dfrac{6}{10}$

Также можно посчитать ответ при помощи формулы комбинаторики:

$P(A) = \dfrac{C_{15}^3}{C_{20}^3} \cdot \dfrac{9}{10} + \dfrac{C_{15}^2 \cdot C_5^1}{C_{20}^3} \cdot \dfrac{8}{10} + \dfrac{C_{15}^1 \cdot C_5^2}{C_{20}^3} \cdot \dfrac{7}{10} + \dfrac{C_5^3}{C_{20}^3} \cdot \dfrac{6}{10}$


```python
first = (sc.comb(15, 3)/sc.comb(20, 3)) * 9/10
second = (sc.comb(15, 2)*sc.comb(5,1)/sc.comb(20, 3)) * 8/10
third = (sc.comb(15, 1)*sc.comb(5, 2)/sc.comb(20, 3)) * 7/10
fourth = (sc.comb(5, 3)/sc.comb(20, 3)) * 6/10
round(first + second + third + fourth, 4)
```

Далее смоделируем процесс вытягивания шара из второй корзины:

```python
import numpy as np

urn1 = np.array([0]*5 + [1]*15)
urn2 = np.array([0] + [1]*6)

rng = np.random.default_rng(42)
N = 1000


result = []
for i in range(N):
  moved = rng.choice(urn1, size=2, replace=False)
  new_urn2 = np.concatenate([urn2, moved])
  ball = rng.choice(new_urn2, size=1)
  result.append(ball)

print(sum(result)/N)
```

```python
import matplotlib.pyplot as plt

result = np.array(result)
freq = np.cumsum(result) / np.arange(1, N + 1)

plt.figure(figsize=(10, 5))
plt.plot(freq, label='Частота белого шара')
plt.axhline(0.825, color='red', linestyle='--', label='Полученное теоретическое значение')
plt.xlabel('Число опытов')
plt.ylabel('Частота')
plt.title('Сходимость частоты к вероятности')
plt.legend()
plt.grid(True)
plt.show()
```
