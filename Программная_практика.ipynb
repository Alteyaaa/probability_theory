# Тема 4. Полная вероятность и формула Байеса

В одной урне 15 белых и 5 черных шаров, а в другой - 6 белых и 1 черный шар. Из первой урны случайным образом вынимают 3 шара и опускают во вторую урну. После этого из второй урны также случайно вынимают 1 шар. Найти вероятность того, что шар, вынутый из второй урны - белый.

* $A$ - вынутый из второй урны шар - белый
* $H_1$ - из первой урны вынули 3 белых шара
* $H_2$ - из первой урны вынули 2 белых шара и 1 черный
* $H_3$ - из первой урны вынули 1 белый шар и 2 черных
* $H_4$ - из первой урны вынули 3 черных шара

Таким образом, вероятности гипотез $H_1 - H_4$:

$P(H_1) = \dfrac{15}{20} \cdot \dfrac{14}{19} \cdot \dfrac{13}{18}$,

$P(H_2) = 3 \cdot \dfrac{15}{20} \cdot \dfrac{14}{19} \cdot \dfrac{5}{18}$,

$P(H_3) = 3 \cdot \dfrac{15}{20} \cdot \dfrac{5}{19} \cdot \dfrac{4}{18}$,

$P(H_4) = \dfrac{5}{20} \cdot \dfrac{4}{19} \cdot \dfrac{3}{18}$.

Множитель 3 в $P(H_2)$ и $P(H_3)$ учитывает, что шары могут быть вынуты в разном порядке.

Найдем условные вероятности:

1.   $P(A|H_1) = \dfrac{9}{10}$
2.   $P(A|H_2) = \dfrac{8}{10}$
3.   $P(A|H_3) = \dfrac{7}{10}$
4.   $P(A|H_4) = \dfrac{6}{10}$

Тогда по формуле полной вероятности:

$P(A) = P(H_1) \cdot P(A|H_1) + P(H_2) \cdot P(A|H_2) + P(H_3) \cdot P(A|H_3) + P(H_4) \cdot P(A|H_4)$





```python
first = 15/20 * 14/19 * 13/18 * 9/10
second = 3 * 15/20 * 14/19 * 5/18 * 8/10
third = 3 * 15/20 * 5/19 * 4/18 * 7/10
fourth = 5/20 * 4/19 * 3/18 * 6/10
round(first + second + third + fourth, 4)
```

Далее смоделируем процесс вытягивания шара из второй урны:

```python
import random

urn1 = [0]*5 + [1]*15      # 0 - черный, 1 - белый
urn2 = [0] + [1]*6

N = 100000
white = 0
freq = []

for i in range(1, N + 1):
  moved = random.sample(urn1, 3)
  new_urn2 = urn2 + moved
  ball = random.choice(new_urn2)
  if ball == 1:
    white += 1
  freq.append(white / i)

print(white / N)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))
plt.plot(freq, label='Частота белого шара')
plt.axhline(0.825, color='red', linestyle='--', label='Полученное теоретическое значение')
plt.xlabel('Число опытов')
plt.ylabel('Частота')
plt.title('Сходимость частоты к вероятности')
plt.legend()
plt.grid(True)
plt.ylim(0.7, 0.9)
plt.show()
```

Рассмотрим событие $A$ - 5 пакетов акций не будут проданы по первоначальной цене.

В таком случае, 4 пакета будут проданы по первоначальной цене, вероятность чего $0.2$, а 5 оставшихся пакетов будут проданы не по первоначальной цене с вероятностью $0.8$. По формуле Бернулли

$P(A) = C_9^5 \cdot 0.8^5 \cdot 0.2^4$

```python
print(round(st.binom.pmf(5, 9, 0.8), 4))
```

Рассмотрим событие $B$ - по первоначально заявленной цене будет продано менее двух пакетов.

Условие говорит о том, что по первоначальной цене будет продан или 1 пакет акций, или 0. Тогда следует рассмотреть 2 вытекающих события отдельно:

$B_1$ - будет продан 1 пакет акций.

$B_2$ - не будет продано ни одного пакета.

Тогда по формуле сложения вероятностей несовместных событий

$P(B) = P(B_1) + P(B_2)$, где по формуле Бернулли

$P(B_1) = C_9^1 \cdot 0.2^1 \cdot 0.8^8$

$P(B_2) = C_9^0 \cdot 0.2^0 \cdot 0.8^9$



```python
first = st.binom.pmf(1, 9, 0.2)
second = st.binom.pmf(0, 9, 0.2)
print(round(first+second, 4))
```

Рассмотрим событие $C$ - по первоначально заявленной цене будет продано не более двух пакетов.

Из условия следует, что будет продано или 2 пакета, или 1, или 0. То есть есть три несовместных события:

$C_1$ - будет продано 2 пакета акций.

$C_2$ - будет продан 1 пакет акций.

$C_3$ - не будет продано ни одного пакета.

$P(C) = P(C_1) + P(C_2) + P(C_3)$

По формуле Бернулли:

$P(С_1) = C_9^2 \cdot 0.2^2 \cdot 0.8^7$

$P(С_2) = C_9^1 \cdot 0.2^1 \cdot 0.8^8$

$P(С_3) = C_9^0 \cdot 0.2^0 \cdot 0.8^9$


```python
first = st.binom.pmf(2, 9, 0.2)
second = st.binom.pmf(1, 9, 0.2)
third = st.binom.pmf(0, 9, 0.2)
print(round(first+second+third, 4))
```

Рассмотрим событие $D$ - по первоначально заявленной цене будет продано хотя бы 2 пакета.

Из условия следует, что по первоначально заявленной цене будет продано 2 и более пакета. Вычислять вручную вероятности каждого события, где проданное число пакетов акций больше или равно двум, долго, поэтому проще посчитать вероятность противоположного события:

$\bar{D}$ - по первоначально заявленной цене будет продано меньше 2 пакетов.

Вероятность данного события мы уже считали, как вероятность события B, поэтому:

$P(D) = 1 - P(\bar{D}) = 1 - P(B)$


```python
first = st.binom.pmf(1, 9, 0.2)
second = st.binom.pmf(0, 9, 0.2)
print(round(1-(first+second), 4))
```

Рассмотрим параметр $m$ - наивероятнейшее число пакетов, проданных по первоначально заявленной цене.

Вычислить это число можно по формуле:

$np-q \leq m \leq np + p$, где

* $p$ - вероятность наступления благоприятной ситуации (у нас 0.2).
* $q$ - противоположная вероятность $p$ (0.8).
* $n$ - общее число испытаний (9).
* $m$ - количество ситуаций из n с благоприятным исходом.

$9 \cdot 0.2 - 0.8 \leq m \leq 9 \cdot 0.2 + 0.2$

$1.8 - 0.8 \leq m \leq 1.8 + 0.2$

$1 \leq m \leq 2$

Следовательно, наивероятнейших чисел два: $m = 1$ и $m = 2$. Тогда вероятность равна

$P = P_{1,9} + P_{2,9} = 0.302 + 0.302 = 0.604$

```python
n, p = 9, 0.2

for m in range(n + 1):
    print(m, round(st.binom.pmf(m, n, p), 4))

print(round(st.binom.pmf(1, n, p) + st.binom.pmf(2, n, p), 4))
```

Рассмотрим событие $A$ - будет повреждено 3 изделия.

Из условия следует, что $m = 3$. Тогда, по формуле Пуассона:

$P(A) = P_3(2) = \dfrac{\lambda^m \cdot e^{-\lambda}}{m!}$

Или, если смотреть по таблице значений $m$ и $\lambda$, то приближенный ответ равен 0.1805.

```python
print(round(st.poisson.pmf(3, 2), 4))
```

Рассмотрим событие $B$ - будет повреждено по крайней мере 3 изделия.

Из условия следует, что будет повреждено $\geq 3$ изделий. Находим через обратную вероятность:

$P(B) = 1 - P(\bar{B})$, где

$P(\bar{B})$ - вероятность того, что будет повреждено меньше 3 изделий (0, 1, 2).

Распишем $P(\bar{B})$ через вероятности повреждения нуля, одного или двух изделий:

$P(\bar{B_0}) = 0.1353$

$P(\bar{B_1}) = 0.2707$

$P(\bar{B_2}) = 0.2707$

Тогда

$P(B) = 1 - (P(\bar{B_0}) + P(\bar{B_1}) + P(\bar{B_2})) = 1 - (0.1353 + 0.2707 \cdot 2) = 0.3233$

```python
first = st.poisson.pmf(0, 2)
second = st.poisson.pmf(1, 2)
third = st.poisson.pmf(2, 2)
print(round(1-(first+second+third), 4))
```

Рассмотрим событие $С$ - не будет повреждено 9997 изделий.

Из условия следует, что повреждено 3 изделия. Вероятность этого события уже была посчитана, как вероятность события $A$.

$P(C) = P(A) = 0.1804$

```python
print(round(st.poisson.pmf(3, 2), 4))
```

Рассмотрим событие $D$ - не будет повреждено хотя бы 9997 изделий.

Из условия следует, что не будет повреждено 9997 и более изделий, т.е. будет повреждено $\leq 3$ изделий. Вероятность этого события можно посчитать как сумму вероятностей повреждения 0, 1, 2 и 3 изделий. Все слагаемые уже были посчитаны ранее, как вероятности $P(\bar{B}) = 0.6767$ (0, 1, 2) и $P(A) = P(C) = 0.1804$ (3). Тогда вероятность равна

$P(D) = P(\bar{B}) + P(A) = 0.6767 + 0.1804 = 0.8571$


```python
first = st.poisson.pmf(0, 2)
second = st.poisson.pmf(1, 2)
third = st.poisson.pmf(2, 2)
fourth = st.poisson.pmf(3, 2)
print(round(first+second+third+fourth, 4))
```

Рассмотрим событие $A$ - 480 предприятий имеют нарушение финансовой дисциплины.

Из условия следует, что $m = 480$. Тогда по локальной теореме Муавра-Лапласа найдем $x$, чтобы воспользоваться табличными значениями.

$x = \dfrac{m - np}{\sqrt{npq}} = \dfrac{480 - 500}{\sqrt{250}} = \dfrac{-20}{\sqrt{250}} = - 1.2649$

Тогда $f(x) = 0.1792$, а вероятность равна

$P(A) = \dfrac{f(x)}{\sqrt{npq}} = \dfrac{0.1792}{\sqrt{250}} = 0.0113$

```python
from math import sqrt
l = 1000 * 0.5
sq = sqrt(1000*0.5*0.5)
print(round(st.norm.pdf(480, loc=l, scale=sq), 4))
```

Рассмотрим событие $B$ - наивероятнейшее число предприятий имеют нарушение финансовой дисциплины.

Наивероятнейшее число $m_0$ найдем из неравенства:

$np - q \le m_0 \le np + p$

$1000 \cdot 0.5 - 0.5 \le m_0 \le 1000 \cdot 0.5 + 0.5$

$499.5 \le m_0 \le 500.5$

Следовательно, $m_0 = 500$. Тогда по локальной теореме Муавра-Лапласа найдем $x$, чтобы воспользоваться табличными значениями.

$x = \frac{m - np}{\sqrt{npq}} = \frac{500 - 500}{\sqrt{250}} = \frac{0}{\sqrt{250}} = 0$

Тогда $f(x) = 0.3989$, а вероятность равна

$P(B) = \frac{f(x)}{\sqrt{npq}} = \frac{0.3989}{\sqrt{250}} = 0.0252$

```python
print(round(st.norm.pdf(500, loc=l, scale=sq), 4))
```

Рассмотрим событие $C$ - не менее 480 предприятий имеют нарушение финансовой дисциплины.

Из условия следует, что $480 \le m \le 1000$. Тогда по интегральной теореме Муавра-Лапласа найдем $x_1$ и $x_2$, чтобы воспользоваться табличными значениями.

$x_1 = \frac{a - np}{\sqrt{npq}} = \frac{480 - 500}{\sqrt{250}} = \frac{-20}{\sqrt{250}} = -1.2649$

$x_2 = \frac{b - np}{\sqrt{npq}} = \frac{1000 - 500}{\sqrt{250}} = \frac{500}{\sqrt{250}} = 31.6228$

Тогда $\Phi(x_2) = \Phi(31.6228) = 1$, $\Phi(x_1) = -\Phi(1.26) = -0.7923$, а вероятность равна

$P(C) = \frac{1}{2}\left[\Phi(x_2) - \Phi(x_1)\right] = \frac{1}{2}\left[1 + 0.7923\right] = 0.8962$

```python
from math import sqrt
l = 1000 * 0.5
sq = sqrt(1000*0.5*0.5)
first = st.norm.cdf(1000, loc=l, scale=sq)
second = st.norm.cdf(480, loc=l, scale=sq)
print(round(first - second, 4))
```

Рассмотрим событие $D$ - от 480 до 520 предприятий имеют нарушение финансовой дисциплины.

Из условия следует, что $480 \le m \le 520$. Границы интервала симметричны относительно $np = 500$, поэтому $|m - 500| \le 20$, то есть $\varepsilon = 20$. Тогда по следствию из интегральной теоремы Муавра-Лапласа найдем $x$, чтобы воспользоваться табличными значениями.

$x = \frac{\varepsilon}{\sqrt{npq}} = \frac{20}{\sqrt{250}} = 1.2649$

Тогда $\Phi(x) = \Phi(1.26) = 0.7923$, а вероятность равна

$P(D) = P(|m - np| \le \varepsilon) = \Phi\left(\frac{\varepsilon}{\sqrt{npq}}\right) = 0.7923$

```python
from math import sqrt
l = 1000 * 0.5
sq = sqrt(1000*0.5*0.5)
first = st.norm.cdf(520, loc=l, scale=sq)
second = st.norm.cdf(480, loc=l, scale=sq)
print(round(first - second, 4))
```
